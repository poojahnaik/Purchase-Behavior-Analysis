```python
import pandas as pd
```

```python
df=pd.read_csv('customer_shopping_behavior_synthetic.csv')
df.head()
```

```python
df.isnull().sum()
```

```python
df['Review Rating']=df.groupby(by='Category')['Review Rating'].transform(lambda x: x.fillna(x.median()))
```

```python
df.isnull().sum()
```

```python
df.columns=df.columns.str.lower()
df.columns=df.columns.str.replace(' ', '_')
df=df.rename(columns={'purchase_amount_(usd)':'purchase_amount'})
```

```python
df.columns
```

```python
df['age'].value_counts().sort_index()
```

```python
labels=['Young Adult', 'Adult', 'Middle-aged', 'Senior']
df['age_group']=pd.qcut(df['age'], 4, labels=labels)
```

```python
df[['age', 'age_group']].head(10)
```

```python
df['frequency_of_purchases'].value_counts()
```

```python
freq_mapping={'Annually':365, 'Every 3 Months':90,
             'Monthly':30, 'Fortnightly':14,
             'Quarterly':90, 'Weekly':7,
             'Bi-Weekly':14}
df['purchase_freq_days']=df['frequency_of_purchases'].map(freq_mapping)
```

```python
df[['frequency_of_purchases', 'purchase_freq_days']].head(10)
```

```python
df[['discount_applied', 'promo_code_used']].head(10)
```

```python
(df['discount_applied']==df['promo_code_used']).all()
```

```python
df=df.drop('promo_code_used', axis=1)
df.head(10)
```
